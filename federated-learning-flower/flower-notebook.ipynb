```python
from mnist_flower.task import load_data
```

```python
trainloader, valloader, testloader = load_data(
    partition_id=0,
    num_partitions=5,
    batch_size=32
)

print(len(trainloader.dataset))
print(len(valloader.dataset))
print(len(testloader.dataset))
```

```python
print("Number of batches:", len(trainloader))
print("Number of training examples:", len(trainloader.dataset))

for index, data in enumerate(trainloader):
    inputs, labels = data
    print("Data shape:", len(data[0]), len(data[1]))
    print("Inputs shape:", inputs.shape)
    print("Labels shape:", labels.shape)
    print("First 10 labels:", labels[:10])
    print("Pixel range:", inputs.min().item(), inputs.max().item())
    break
```
